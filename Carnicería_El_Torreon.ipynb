```python
# %% [markdown]
"""
# Sistema Predictivo para Optimización de Recursos - Carnicería "El Torreón"

Este notebook simula el análisis predictivo desarrollado para la carnicería "El Torreón", mostrando comparativas entre la situación antes y después de implementar el sistema.

"""
# %% [markdown]
"""
## 1. Configuración inicial
Importamos las bibliotecas necesarias y configuramos el entorno
"""

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import random

# Configuración de estilo
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (12, 6)
sns.set_palette("husl")

# %% [markdown]
"""
## 2. Generación de datos ficticios
Crearemos datos históricos simulados para 6 meses antes y después de implementar el sistema
"""

# %%
# Configuración de parámetros
productos = ['Sirloin', 'Arrachera', 'Chuleta', 'Pollo entero', 'Costilla', 'Molida']
dias_semana = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
sucursales = ['Centro', 'Villa', 'Jardín']

# Función para generar datos ficticios (situación inicial sin sistema)
def generar_datos_iniciales(fecha_inicio, duracion_meses):
    rango_fechas = pd.date_range(fecha_inicio, periods=duracion_meses*30, freq='D')
    datos = []

    for fecha in rango_fechas:
        for sucursal in sucursales:
            dia_semana = dias_semana[fecha.weekday()]

            # Variabilidad por día de la semana
            if dia_semana in ['Viernes', 'Sábado']:
                factor_demanda = random.uniform(1.5, 2.0)
            elif dia_semana == 'Domingo':
                factor_demanda = random.uniform(1.2, 1.6)
            else:
                factor_demanda = random.uniform(0.7, 1.2)

            for producto in productos:
                # Base de ventas por producto
                if producto == 'Sirloin':
                    ventas = int(random.normalvariate(25, 5) * factor_demanda)
                elif producto == 'Arrachera':
                    ventas = int(random.normalvariate(20, 4) * factor_demanda)
                elif producto == 'Pollo entero':
                    ventas = int(random.normalvariate(15, 3) * factor_demanda)
                else:
                    ventas = int(random.normalvariate(12, 3) * factor_demanda)

                # Pérdidas estimadas (merma + robos)
                perdidas = int(ventas * random.uniform(0.05, 0.15))  # 5-15% de pérdidas

                datos.append({
                    'Fecha': fecha,
                    'Dia_Semana': dia_semana,
                    'Sucursal': sucursal,
                    'Producto': producto,
                    'Ventas': ventas,
                    'Perdidas': perdidas,
                    'Stock_Inicial': int(ventas * random.uniform(1.5, 2.5)),  # Sobrestock común
                    'Stock_Final': 0  # Se calculará después
                })

    df = pd.DataFrame(datos)

    # Calcular stock final (simulando mala gestión)
    df['Stock_Final'] = df.apply(lambda x: max(0, x['Stock_Inicial'] - x['Ventas'] - x['Perdidas']), axis=1)

    return df

# Función para generar datos después de implementar el sistema
def generar_datos_mejorados(fecha_inicio, duracion_meses, df_historico):
    rango_fechas = pd.date_range(fecha_inicio, periods=duracion_meses*30, freq='D')
    datos = []

    # Analizar patrones históricos para mejorar predicciones
    patron_semanal = df_historico.groupby(['Dia_Semana', 'Producto'])['Ventas'].mean().reset_index()

    for fecha in rango_fechas:
        for sucursal in sucursales:
            dia_semana = dias_semana[fecha.weekday()]

            for producto in productos:
                # Obtener promedio histórico para este día y producto
                ventas_promedio = patron_semanal[
                    (patron_semanal['Dia_Semana'] == dia_semana) &
                    (patron_semanal['Producto'] == producto)
                ]['Ventas'].values[0]

                # Variación alrededor del promedio (más ajustada que antes)
                ventas = int(random.normalvariate(ventas_promedio, ventas_promedio*0.1))

                # Pérdidas reducidas por mejor control
                perdidas = int(ventas * random.uniform(0.02, 0.05))  # 2-5% de pérdidas

                # Stock optimizado usando predicción
                stock_inicial = int(ventas * random.uniform(1.1, 1.3))  # Menos sobrestock

                datos.append({
                    'Fecha': fecha,
                    'Dia_Semana': dia_semana,
                    'Sucursal': sucursal,
                    'Producto': producto,
                    'Ventas': ventas,
                    'Perdidas': perdidas,
                    'Stock_Inicial': stock_inicial,
                    'Stock_Final': max(0, stock_inicial - ventas - perdidas)
                })

    return pd.DataFrame(datos)

# Generar datos
fecha_inicio = '2024-07-01'
df_inicial = generar_datos_iniciales(fecha_inicio, 6)
df_mejorado = generar_datos_mejorados('2025-01-01', 6, df_inicial)

# %% [markdown]
"""
## 3. Análisis Comparativo: Situación Antes vs. Después

### 3.1. Pérdidas por merma y robos
"""

# %%
# Calcular pérdidas totales antes y después
perdidas_antes = df_inicial.groupby('Sucursal')['Perdidas'].sum()
perdidas_despues = df_mejorado.groupby('Sucursal')['Perdidas'].sum()

# Gráfico comparativo
plt.figure(figsize=(10, 6))
bar_width = 0.35
index = np.arange(len(sucursales))

bars1 = plt.bar(index, perdidas_antes, bar_width, label='Antes del sistema', alpha=0.8)
bars2 = plt.bar(index + bar_width, perdidas_despues, bar_width, label='Con sistema predictivo', alpha=0.8)

plt.xlabel('Sucursal')
plt.ylabel('Total de pérdidas (unidades)')
plt.title('Comparación de Pérdidas por Merma/Robos: Antes vs. Después')
plt.xticks(index + bar_width / 2, sucursales)
plt.legend()

# Añadir valores exactos
for bar in bars1 + bars2:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             '%d' % int(height),
             ha='center', va='bottom')

plt.tight_layout()
plt.show()

# %% [markdown]
"""
### 3.2. Eficiencia en gestión de inventario
"""

# %%
# Calcular porcentaje de sobrestock antes y después
df_inicial['Sobrestock'] = df_inicial['Stock_Final'] / df_inicial['Stock_Inicial']
df_mejorado['Sobrestock'] = df_mejorado['Stock_Final'] / df_mejorado['Stock_Inicial']

# Gráfico de boxplot comparativo
plt.figure(figsize=(12, 6))
sns.boxplot(x='Sucursal', y='Sobrestock', hue='Periodo',
            data=pd.concat([
                df_inicial.assign(Periodo='Antes'),
                df_mejorado.assign(Periodo='Después')
            ]), palette="Set2")
plt.title('Distribución del Sobrestock: Antes vs. Después')
plt.ylabel('Porcentaje de Stock no vendido')
plt.axhline(y=0.1, color='r', linestyle='--', alpha=0.5)
plt.text(2.5, 0.11, 'Objetivo ideal', color='r')
plt.tight_layout()
plt.show()

# %% [markdown]
"""
### 3.3. Ventas por día de la semana
"""

# %%
# Preparar datos para comparación
ventas_semana_antes = df_inicial.groupby('Dia_Semana')['Ventas'].sum().reindex(dias_semana)
ventas_semana_despues = df_mejorado.groupby('Dia_Semana')['Ventas'].sum().reindex(dias_semana)

# Gráfico de líneas comparativo
plt.figure(figsize=(12, 6))
plt.plot(dias_semana, ventas_semana_antes, marker='o', label='Antes del sistema', linewidth=2)
plt.plot(dias_semana, ventas_semana_despues, marker='o', label='Con sistema predictivo', linewidth=2)

plt.title('Comparación de Ventas por Día de la Semana')
plt.xlabel('Día de la semana')
plt.ylabel('Total de ventas (unidades)')
plt.grid(True, alpha=0.3)
plt.legend()

# Añadir valores de mejora porcentual
for i, (antes, despues) in enumerate(zip(ventas_semana_antes, ventas_semana_despues)):
    mejora = (despues - antes) / antes * 100
    plt.text(i, max(antes, despues) + 50, f'{mejora:.1f}%', ha='center')

plt.tight_layout()
plt.show()

# %% [markdown]
"""
### 3.4. Rotación de productos
"""

# %%
# Calcular rotación antes y después
rotacion_antes = df_inicial.groupby('Producto')['Ventas'].sum().sort_values(ascending=False)
rotacion_despues = df_mejorado.groupby('Producto')['Ventas'].sum().sort_values(ascending=False)

# Gráfico de barras comparativo
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Rotación de Productos: Comparación Antes vs. Después')

rotacion_antes.plot(kind='bar', ax=ax1, color='skyblue')
ax1.set_title('Antes del sistema')
ax1.set_ylabel('Total de ventas (unidades)')
ax1.tick_params(axis='x', rotation=45)

rotacion_despues.plot(kind='bar', ax=ax2, color='lightgreen')
ax2.set_title('Con sistema predictivo')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# %% [markdown]
"""
### 3.5. Predicción de demanda (ejemplo para Sirloin)
"""

# %%
# Filtrar datos para Sirloin en sucursal Centro
sirloin_centro_antes = df_inicial[
    (df_inicial['Producto'] == 'Sirloin') &
    (df_inicial['Sucursal'] == 'Centro')
].groupby('Fecha')['Ventas'].sum()

sirloin_centro_despues = df_mejorado[
    (df_mejorado['Producto'] == 'Sirloin') &
    (df_mejorado['Sucursal'] == 'Centro')
].groupby('Fecha')['Ventas'].sum()

# Calcular promedios móviles (simulando predicción)
ventana = 7  # Promedio semanal
sirloin_centro_antes_ma = sirloin_centro_antes.rolling(ventana).mean()
sirloin_centro_despues_ma = sirloin_centro_despues.rolling(ventana).mean()

# Gráfico de series temporales
plt.figure(figsize=(14, 6))
plt.plot(sirloin_centro_antes.index, sirloin_centro_antes, label='Ventas reales (antes)', alpha=0.5)
plt.plot(sirloin_centro_antes.index, sirloin_centro_antes_ma,
         label=f'Predicción (promedio {ventana} días)', color='red', linewidth=2)
plt.title('Demanda de Sirloin en sucursal Centro - Antes del sistema')
plt.xlabel('Fecha')
plt.ylabel('Ventas diarias (unidades)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(14, 6))
plt.plot(sirloin_centro_despues.index, sirloin_centro_despues, label='Ventas reales (después)', alpha=0.5)
plt.plot(sirloin_centro_despues.index, sirloin_centro_despues_ma,
         label=f'Predicción optimizada ({ventana} días)', color='green', linewidth=2)
plt.title('Demanda de Sirloin en sucursal Centro - Con sistema predictivo')
plt.xlabel('Fecha')
plt.ylabel('Ventas diarias (unidades)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# %% [markdown]
"""
## 4. Resumen de métricas clave
"""

# %%
# Calcular métricas comparativas
def calcular_metricas(df):
    return {
        'Perdidas_Promedio': df['Perdidas'].mean(),
        'Sobrestock_Promedio': df['Sobrestock'].mean(),
        'Rotacion_Total': df['Ventas'].sum(),
        'Dias_Agotado': (df['Stock_Final'] == 0).sum() / len(df) * 100,
        'Eficiencia_Inventario': df['Ventas'].sum() / df['Stock_Inicial'].sum() * 100
    }

metricas_antes = calcular_metricas(df_inicial)
metricas_despues = calcular_metricas(df_mejorado)

# Crear tabla comparativa
df_metricas = pd.DataFrame({
    'Antes del sistema': metricas_antes,
    'Con sistema predictivo': metricas_despues,
    'Mejora': [(metricas_despues[k] - metricas_antes[k]) / metricas_antes[k] * 100
               for k in metricas_antes]
}).T

# Formatear para mejor visualización
df_metricas['Perdidas_Promedio'] = df_metricas['Perdidas_Promedio'].round(1)
df_metricas['Sobrestock_Promedio'] = (df_metricas['Sobrestock_Promedio'] * 100).round(1).astype(str) + '%'
df_metricas['Rotacion_Total'] = df_metricas['Rotacion_Total'].astype(int)
df_metricas['Dias_Agotado'] = df_metricas['Dias_Agotado'].round(1).astype(str) + '%'
df_metricas['Eficiencia_Inventario'] = df_metricas['Eficiencia_Inventario'].round(1).astype(str) + '%'
df_metricas.loc['Mejora', ['Sobrestock_Promedio', 'Dias_Agotado', 'Eficiencia_Inventario']] = '-'

# Mostrar tabla
df_metricas.style\
    .set_caption('Métricas Clave: Comparación Antes vs. Después')\
    .set_properties({'text-align': 'center'})\
    .highlight_max(axis=0, color='lightgreen')\
    .highlight_min(axis=0, color='lightcoral')

# %% [markdown]
```
